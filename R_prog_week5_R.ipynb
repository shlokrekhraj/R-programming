```python
%%R
# Assign NA
x = NA
x

# Check missing value
is.na(x)

# Vector with missing values
x = c(11, NA, 13, NA)
is.na(x)

# Mean with NA
mean(x)

# Remove NA while calculating mean
mean(x, na.rm = TRUE)
```

```python
%%R
# Locate missing values
x = c(11, NA, 13, NA)
which(is.na(x))

# Count missing values
sum(is.na(x))

# Find complete cases
complete.cases(x)

# Remove NA values
y = na.omit(x)
y

# Mean after removing NA
mean(x)
mean(y)
```
