```r
cat("========== LECTURE 26: Sequences ==========\n")

# Continuous sequences
print(1:10)
print(10:1)
print(5:15)
print(15:5)
print(-1:-10)
print(-10:-1)
print(-5:-15)
print(-15:-5)

# Decimal sequences
print(1.23:10)
print(1.23:10.54)
print(10.54:2.23)
print(-1.23:-10)
print(-5.23:6)

# seq() function
print(seq(10))
print(seq(1:10))

# Variable sequences
x = 2
print(seq(1, x, x/10))

x = 50
print(seq(0, x, x/10))

# Store sequences
x = seq(1, 50, 1/2)
print(x)

y = 2 * x
print(y)

# Index vectors
x = c(9, 8, 7, 6)
ind = seq(along=x)
print(ind)
print(x[ind[2]])

# Time & Date
print(Sys.time())
print(Sys.Date())


cat("\n========== LECTURE 27: Dates & Alphabets ==========\n")

print(seq(as.Date("2010-01-01"), as.Date("2017-01-01"), by="years"))
print(seq(as.Date("2017-01-01"), by="days", length=6))
print(seq(as.Date("2017-01-01"), by="months", length=6))
print(seq(as.Date("2017-01-01"), by="years", length=6))

startdate = as.Date("2016-01-01")
enddate = as.Date("2017-01-01")
out = seq(enddate, startdate, by="-1 month")
print(out)

# Alphabets
print(letters)
print(letters[1:3])
print(letters[3:1])
print(letters[21:23])
print(letters[2])

print(LETTERS)
print(LETTERS[1:3])
print(LETTERS[3:1])
print(LETTERS[21:23])
print(LETTERS[2])


cat("\n========== LECTURE 28: Repeats ==========\n")

print(rep(3.5, times=10))
print(rep(1:4, 2))

x = 1:4
print(rep(x, times=3))
print(rep(x, each=3))

print(rep(1:4, each=2))
print(rep(1:4, times=3, each=2))

print(rep(1:4, 2:5))

ans = seq(2, 8, 2)
print(ans)
print(rep(1:4, ans))

# Matrix repeat
x = matrix(1:4, nrow=2, byrow=TRUE)
print(x)
print(rep(x, 3))

# Character repeat
print(rep(c("a","b","c"),2))
print(rep(c("apple","banana","cake"),2))
print(rep(2, length.out=5))
print(rep(c(2,3), length.out=5))
print(rep(c(2,3,4), length.out=5))
print(rep("apple", length.out=5))
print(rep(c("a","b","c"), length.out=2))
print(rep(c("a","b","c"), length.out=5))


cat("\n========== LECTURE 29: Sorting & Mode ==========\n")

y = c(8,5,7,6)
print(y)
print(sort(y))
print(sort(y, decreasing=TRUE))

y = c(9,8,5,7,6)
print(order(y))
print(order(y, decreasing=TRUE))

print(mode(2.432))
print(mode(c(3,4,5)))
print(mode("India"))
print(mode(c("India","CANADA")))
print(mode(factor(c("UP","MP"))))
print(mode(list("India","USA")))
print(mode(data.frame(x=1:2,y=c("India","USA"))))
print(mode(print))


cat("\n========== LECTURE 30: Lists ==========\n")

x1 = matrix(1:4, nrow=2, byrow=TRUE)
x2 = matrix(5:8, nrow=2, byrow=TRUE)

print(x1)
print(x2)
print(x1 + x2)

# ⚠️ Avoid breaking execution: show but don't use in addition
x1_temp = x1
x1_temp[2,1] = "hello"
print(x1_temp)
cat("Note: Cannot add x1_temp with x2 due to type change\n")

# Lists
x1 = matrix(1:4, nrow=2, byrow=TRUE)
x2 = matrix(5:8, nrow=2, byrow=TRUE)

matlist = list(x1, x2)
print(matlist)

print(matlist[1])     # sublist
print(matlist[2])

print(matlist[[1]])   # actual matrix
print(matlist[[2]])

# Mixed list
z1 = list(
  c("water","juice","lemonade"),
  rep(1:4, each=2),
  matrix(5:8, nrow=2, byrow=TRUE)
)

print(z1)

# Extraction
print(z1[[1]])
print(z1[1][2])      # NULL (wrong indexing)
print(z1[[1]][2])    # correct
```
